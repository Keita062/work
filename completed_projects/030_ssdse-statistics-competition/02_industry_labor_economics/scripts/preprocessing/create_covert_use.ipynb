```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import japanize_matplotlib
import os
```

```python
# ファイルパスの定義
file_wage = r"C:\GitHub\Data-Analysis_competition\SSDSE\02_industry_labor_economics\Data\USE\勤続年数・超過実労働時間数・所定内給与額_時系列.csv"
file_bankruptcy = r"C:\GitHub\Data-Analysis_competition\SSDSE\02_industry_labor_economics\Data\USE\産業別年間倒産件数表.csv"
file_unemployed = r"C:\GitHub\Data-Analysis_competition\SSDSE\02_industry_labor_economics\Data\USE\失業者 - シート1.csv"


# データの読み込み
df_wage = pd.read_csv(file_wage)
df_wage = df_wage.sort_values('年度')
df_bankruptcy = pd.read_csv(file_bankruptcy)
df_unemployed = pd.read_csv(file_unemployed)
```

```python
print("勤続年数・超過実労働時間数・所定内給与額の基本情報:", df_wage.info())
```

```python
print("産業別年間倒産件数表の基本情報:", df_bankruptcy.info())
```

```python
print("失業者の基本情報:", df_unemployed.info())
```

```python
print("産業年間倒産件数の「年」のユニーク値:\n", df_bankruptcy['年'].unique())
print("\n")
print("失業者の「年度」のユニーク値:\n", df_unemployed['年度'].unique())
print("\n")
print("勤続年数・超過実労働時間数・所定内給与額の「年度」のユニーク値:\n", df_wage['年度'].unique())
```

```python
# 失業者データ (df_unemployed) で「年度」が nan になっている行を抽出して表示
# ※もしエラーが出る場合は、カラム名を '年度' に変更してください
df_unemployed[df_unemployed['年度'].isnull()]
# ※ もし KeyError が出る場合は '年' に書き換えてください
col_unemployed = '年度' 

# nan が含まれる行を削除（元の df_unemployed を上書き更新します）
df_unemployed = df_unemployed.dropna(subset=[col_unemployed])
print("失業者の「年度」のユニーク値:\n", df_unemployed['年度'].unique())
```

```python
# ① 産業年間倒産件数 (df_bankruptcy)
# 「2014年」などの文字列から「年」を削除して整数型に変換
df_bankruptcy['年_共通'] = df_bankruptcy['年'].str.replace('年', '').astype(int)

# ② 失業者 (df_unemployed)
# nanを削除したため、安全に少数点（.0）のない整数型(int)に変換できます
df_unemployed['年_共通'] = df_unemployed[col_unemployed].astype(int)

# ③ 勤続年数・給与 (df_wage)
# 和暦を西暦（整数）に変換するマッピング
wareki_to_seireki = {
    '平成18年': 2006, '平成19年': 2007, '平成20年': 2008, '平成21年': 2009, '平成22年': 2010,
    '平成23年': 2011, '平成24年': 2012, '平成25年': 2013, '平成26年': 2014, '平成27年': 2015,
    '平成28年': 2016, '平成29年': 2017, '平成30年': 2018,
    '令和元年': 2019, '令和二年': 2020, '令和三年': 2021, '令和四年': 2022, '令和五年': 2023,
    '令和六年': 2024, '令和七年': 2025
}
df_wage['年_共通'] = df_wage['年度'].map(wareki_to_seireki)
```

```python
print("① 倒産件数 (df_bankruptcy) の『年_共通』:", sorted(df_bankruptcy['年_共通'].unique()))
print("② 失業者   (df_unemployed) の『年_共通』:", sorted(df_unemployed['年_共通'].unique()))
print("③ 賃金・給与 (df_wage)     の『年_共通』:", sorted(df_wage['年_共通'].unique()))
```
