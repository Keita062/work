```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import japanize_matplotlib
```

```python
base_dir = r"C:\GitHub\Data-Analysis_competition\SSDSE\02_industry_labor_economics\Data\covert_use"

df_wage = pd.read_csv(os.path.join(base_dir, "勤続年数・超過実労働時間数・所定内給与額_時系列.csv"))
df_bankruptcy = pd.read_csv(os.path.join(base_dir, "産業別年間倒産件数表.csv"))
df_unemployed = pd.read_csv(os.path.join(base_dir, "失業者 - シート1.csv"))
```

```python
# 各データの「年_共通」列のユニーク値を集合（set）にする
years_wage = set(df_wage['年_共通'].unique())
years_bankruptcy = set(df_bankruptcy['年_共通'].unique())
years_unemployed = set(df_unemployed['年_共通'].unique())

common_years = sorted(list(years_wage & years_bankruptcy & years_unemployed))
```

```python
common_years
```

```python
df_wage_filtered = df_wage[df_wage['年_共通'].isin(common_years)]
df_bankruptcy_filtered = df_bankruptcy[df_bankruptcy['年_共通'].isin(common_years)]
df_unemployed_filtered = df_unemployed[df_unemployed['年_共通'].isin(common_years)]
```

```python
base_dir_1 = r"C:\GitHub\Data-Analysis_competition\SSDSE\02_industry_labor_economics\Data\covert_use_fileted"
```

```python
df_wage_filtered.to_csv(os.path.join(base_dir_1, "勤続年数・超過実労働時間数・所定内給与額_時系列_共通年.csv"), index=False, encoding='utf-8-sig')
df_bankruptcy_filtered.to_csv(os.path.join(base_dir_1, "産業別年間倒産件数表_共通年.csv"), index=False, encoding='utf-8-sig')
df_unemployed_filtered.to_csv(os.path.join(base_dir_1, "失業者 - シート1_共通年.csv"), index=False, encoding='utf-8-sig')
```

```python
df_unemployed['前職の産業'] = (
    df_unemployed['前職の産業']
    .astype(str)
    .str.replace('\n', '')
    .str.replace(' ', '')
    .str.replace('　', '')
    .str.replace('(', '（')
    .str.replace(')', '）')
)
df_unemployed['前職の産業'].unique()
```

```python
target_cols = ['男女計（失業者数）', '男（失業者数）', '女（失業者数）']
for col in target_cols:
    df_unemployed[col] = df_unemployed[col].astype(str).str.strip()
    # replace('-', np.nan) を行うことで、0とは明確に区別された欠損値になります
    df_unemployed[col] = pd.to_numeric(df_unemployed[col].replace('-', np.nan), errors='coerce')
```
