```python
# グラフ全体のデザインスタイルを設定
sns.set_theme(style="whitegrid")

# 1. データの読み込みパスと保存先パスの設定（raw文字列として定義）
input_dir = r"C:\GitHub\Data-Analysis_competition\SSDSE\02_industry_labor_economics\Data\covert_use_fileted_update_1"
save_dir = r"C:\GitHub\Data-Analysis_competition\SSDSE\02_industry_labor_economics\Data\shitugyoushasu_時系列"

# 保存先フォルダが存在しない場合は自動作成
if not os.path.exists(save_dir):
    os.makedirs(save_dir)
    
# データの読み込み
df_shitsugyosha = pd.read_csv(os.path.join(input_dir, "失業者 - シート1_共通年_統一.csv"))

# 2. 「産業合計」を除外した、個別産業の一覧を取得
industries = [ind for ind in df_shitsugyosha['共通産業列'].unique() if ind != '産業合計']

# 3. forループで産業ごとにグラフを描画・保存
for industry in industries:
    
    # 該当する産業のデータだけを抽出し、年の昇順に並び替え
    df_target = df_shitsugyosha[df_shitsugyosha['共通産業列'] == industry].sort_values('年_共通')
    
    # ループごとに新しいグラフを作成
    plt.figure(figsize=(10, 5))
    
    # ① 男女計の折れ線（太い青線）
    sns.lineplot(
        data=df_target, x='年_共通', y='男女計（失業者数）', 
        marker='o', linewidth=3, color='#1f77b4', label='男女計'
    )
    
    # ② 男の折れ線（薄い青の点線）
    sns.lineplot(
        data=df_target, x='年_共通', y='男（失業者数）', 
        marker='s', linewidth=1.5, color='#aec7e8', linestyle='--', label='男'
    )
    
    # ③ 女の折れ線（オレンジの点線）
    sns.lineplot(
        data=df_target, x='年_共通', y='女（失業者数）', 
        marker='^', linewidth=1.5, color='#ff7f0e', linestyle='--', label='女'
    )
    
    # グラフのタイトルやラベルの設定
    plt.title(f'【時系列推移】{industry} の失業者数推移', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel('年（西暦）', fontsize=11)
    plt.ylabel('失業者数', fontsize=11)
    
    # 横軸・y軸の制約を設定（0〜60、10刻み）
    plt.xticks(df_target['年_共通'].unique())
    plt.ylim(0, 60)
    plt.yticks(range(0, 61, 10))
    
    # 凡例の位置
    plt.legend(title='内訳', loc='upper right')
    plt.tight_layout()
    
   # ファイル名からスラッシュなどの禁止文字を除去（念のための対策）
    clean_industry_name = industry.replace('/', '_').replace('\\', '_')
    file_name = f"{clean_industry_name}_失業者数推移.png"
    file_path = os.path.join(save_dir, file_name)
    
    # bbox_inches='tight' で文字切れを防ぎ、dpi=150 で綺麗な画質で保存
    plt.savefig(file_path, bbox_inches='tight', dpi=150)
    
    # 画面への表示
    plt.show()
```
